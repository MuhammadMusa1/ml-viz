## Три главных слова: датасет, признак, целевая переменная

Датасет (dataset) — это таблица с данными, набор примеров для обучения. Каждая строка — один объект (например, один пингвин, одна квартира, одно письмо). Каждый столбец — одна характеристика объекта, её называют признаком (feature): длина клюва, масса, цена, площадь.

Среди столбцов часто есть один особенный — тот, который мы хотим предсказывать. Его называют целевой переменной (target). Например, «какой вид у пингвина». Все остальные признаки — это подсказки, по которым модель будет угадывать целевую переменную. Разделять признаки и цель важно: именно связь «признаки → цель» и ищет модель.

## Первый код: открываем датасет

Загрузим известный учебный датасет о пингвинах Палмера — мы будем возвращаться к нему на протяжении нескольких глав, поэтому стоит с ним подружиться. Вставь этот код в новую ячейку Google Colab и запусти (Shift+Enter).

```python
# Импортируем библиотеку Pandas под коротким именем pd
# (pd — общепринятое сокращение, его используют все)
import pandas as pd

# Загружаем датасет о пингвинах Палмера прямо из интернета по ссылке
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url).drop(columns=["sex"])   # df — DataFrame; колонку sex убираем

# Показываем первые 5 строк таблицы
df.head()
```

_Ожидаемый результат: таблица из 5 строк с колонками species, island, bill_length_mm, flipper_length_mm, body_mass_g и другими._

Разберём, что произошло. pd.read_csv читает файл формата CSV (таблица, где значения разделены запятыми) и превращает его в DataFrame — главный объект Pandas, ту самую «таблицу внутри Python». Мы сохранили её в переменную df и теперь можем с ней работать.

## Фильтрация и группировка: привет из SQL

Если ты знаком с SQL, следующие операции покажутся родными. Pandas умеет то же самое, что запросы к базе данных, только внутри Python. Даже если SQL ты не знаешь — ничего страшного, разберёмся с нуля.

### Фильтрация — отбор строк по условию (как WHERE)

```python
# Оставить только пингвинов вида Adelie
adelie = df[df["species"] == "Adelie"]
adelie.head()

# Оставить только лёгких пингвинов (меньше 3500 г)
light = df[df["body_mass_g"] < 3500]
print("Лёгких пингвинов:", len(light))
```

Первый пример — прямой аналог SQL-запроса SELECT * FROM df WHERE species = 'Adelie'. Внутри квадратных скобок мы пишем условие, и остаются только строки, для которых оно истинно. Условия можно комбинировать: df[(df["body_mass_g"] < 3500) & (df["species"] == "Adelie")] — лёгкие пингвины вида Adelie. Знак & означает «и».

**Внимание, частая ошибка  ⚠️**

Легко перепутать одиночное = (присваивание значения) и двойное == (сравнение). В условии фильтра всегда двойное: df["species"] == "Adelie". Одиночное = здесь вызовет ошибку. Ещё частая ошибка — забыть скобки при комбинировании условий: пиши (df["body_mass_g"] < 3500) & (df["species"] == "Adelie"), каждое условие в своих скобках.

### Группировка — сводка по категориям (как GROUP BY)

```python
# Средняя масса тела отдельно для каждого вида
df.groupby("species")["body_mass_g"].mean()

# Средняя длина ласта по островам
df.groupby("island")["flipper_length_mm"].mean()
```

Первая строка отвечает на настоящий вопрос: у какого вида пингвины крупнее? Результат сразу бросается в глаза — пингвины вида Gentoo заметно тяжелее (около 5000 г против ~3700 г у остальных). Это и есть паттерн в данных, найденный за секунду одной строкой кода. Группировка — один из самых мощных инструментов быстрого анализа: она сворачивает тысячи строк в короткую понятную сводку.

## Сортировка — что в топе

```python
# 5 самых тяжёлых пингвинов
df.sort_values("body_mass_g", ascending=False).head(5)
```

Сортировка помогает найти крайние случаи: самых тяжёлых пингвинов, самых лёгких, с самым длинным клювом. Часто именно в крайностях прячется что-то интересное или ошибочное.
