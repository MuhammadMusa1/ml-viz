## Проблема 1: пропущенные значения

Реальные данные почти всегда неполные. Кто-то не указал возраст, где-то потерялась запись, датчик не сработал. В данных о пингвинах у 2 особей не измерена масса тела. Модель не умеет работать с пустыми клетками — она выдаст ошибку. Поэтому пропуски обязательно нужно как-то обработать перед обучением.

```python
# Данные о пингвинах (выполни, если открыл новый блокнот)
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url).drop(columns=["sex"])

# Сколько пропусков в каждой колонке?
df.isnull().sum()

# --- Вариант 1: удалить строки с пропусками ---
# Просто, но осторожно: теряем данные целыми строками
df_clean = df.dropna()

# --- Вариант 2: заполнить пропуски (чаще лучше) ---
# Заполняем медианой — серединным значением
df["body_mass_g"] = df["body_mass_g"].fillna(df["body_mass_g"].median())
```

Какой вариант выбрать? Если пропусков мало (пара процентов) — можно удалить строки. Если пропусков много, удаление выбросит слишком много данных, и лучше заполнить. Для чисел заполняют средним или медианой, для категорий — самым частым значением.

**Полезно знать  ℹ️**

Заполнять пропуски медианой (серединным значением), а не средним, обычно безопаснее: медиана не «съезжает» из-за нескольких очень больших или очень маленьких значений. Если в данных о зарплатах есть один миллиардер, среднее взлетит и исказит картину, а медиана останется реалистичной.

## Проблема 2: выбросы

Выброс (outlier) — это значение, которое сильно выбивается из общего ряда. Например, если у одного пингвина указана масса в 10 раз больше остальных, или длина клюва 500 мм. Выбросы бывают двух видов: ошибки (опечатка при вводе — их нужно исправить или убрать) и настоящие редкие случаи (реально очень крупная особь — их важно сохранить). Найти выбросы помогает график «ящик с усами» (boxplot).

```python
import matplotlib.pyplot as plt

# Ящик с усами по массе тела
# Точки далеко за "усами" — потенциальные выбросы
df.boxplot(column="body_mass_g")
plt.title("Поиск выбросов в массе пингвинов")
plt.show()
```

Как читать boxplot: «ящик» — это основная масса данных (средние 50%), «усы» тянутся к типичным крайним значениям, а отдельные точки за усами — кандидаты в выбросы. Увидев выброс, задай вопрос: это ошибка или реальность? От ответа зависит, убирать его или оставить.
