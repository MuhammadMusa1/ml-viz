**Аналогия: ученик, исправляющий ошибки  💡**

Нейросеть учится, как школьник, который решает задачу, сверяется с ответом в конце учебника, видит, где ошибся, и чуть-чуть корректирует свой подход. Потом снова, и снова, тысячи раз. С каждым разом ошибок всё меньше. Никто не даёт сети готовых правил — она сама постепенно подстраивается под данные, шаг за шагом уменьшая ошибку. Терпеливое исправление ошибок — вот и весь секрет.

## Нейрон — это взвешенная сумма

Искусственный нейрон гораздо проще биологического. Он берёт несколько чисел на вход, умножает каждое на свой вес (число, показывающее, насколько этот вход важен), складывает всё вместе и пропускает результат через функцию, которая решает, «сработать» нейрону или нет. Вот и всё устройство одного нейрона — сложение с весами.

**Аналогия: приёмная комиссия  💡**

Нейрон — как член приёмной комиссии, который взвешивает факторы об абитуриенте: оценки × 0,5, олимпиады × 0,3, собеседование × 0,2. Он складывает взвешенные баллы и выносит решение. Вес показывает, насколько для него важен каждый фактор: чем больше вес, тем сильнее фактор влияет на итог. Обучение сети — это и есть подбор правильных весов для всех нейронов.

## Слои: от простого к сложному

Один нейрон умеет мало. Но если выстроить нейроны в слои — где выход одного слоя становится входом следующего — сеть научается распознавать всё более сложные вещи. Первый слой ловит простые детали (края, точки, наклоны), следующие складывают из них узоры (углы, дуги), а последний выдаёт итоговый ответ. Чем больше слоёв, тем «глубже» сеть — отсюда и название «глубокое обучение».

Слой

Роль

Входной

Принимает данные (например, пиксели картинки или числа-признаки)

Скрытые

Находят закономерности, от простых к сложным; их может быть много

Выходной

Выдаёт итоговый ответ (например, «это цифра 7» или вероятность)

## Как сеть учится: подкрутить веса

Сеть делает предсказание, сравнивает его с правильным ответом и вычисляет ошибку (насколько промахнулась). Затем специальный алгоритм — обратное распространение (backpropagation) — определяет, какие именно веса виноваты в ошибке и в какую сторону их чуть подправить, чтобы в следующий раз ошибиться меньше. Один полный проход по всем обучающим данным называется эпохой (epoch). Обучение — это много эпох подряд: сеть снова и снова видит данные и всё точнее подкручивает веса.

**Полезно знать  ℹ️**

Не нужно понимать математику обратного распространения, чтобы пользоваться нейросетями, — библиотека Keras делает всё сама. Достаточно интуиции: сеть измеряет ошибку и чуть-чуть двигает веса в сторону её уменьшения, повторяя это тысячи раз. Как спуск с горы на ощупь: маленький шаг вниз, ещё шаг, ещё — пока не окажешься в низине.

## Первая нейросеть на Keras

Keras — библиотека, которая превращает создание нейросетей почти в конструктор. Сеть собирается как башня из кубиков-слоёв. Инструмент Sequential — это стопка слоёв друг за другом, а Dense — обычный полносвязный слой, где каждый нейрон связан со всеми нейронами предыдущего слоя.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Стопка слоёв: 8 нейронов -> 8 нейронов -> 1 ответ
model = Sequential([
    Dense(8, activation="relu", input_shape=(4,)),  # входной слой, 4 признака
    Dense(8, activation="relu"),                     # скрытый слой
    Dense(1, activation="sigmoid"),                  # выход: число 0..1
])

# Настраиваем, КАК учиться
model.compile(optimizer="adam", loss="binary_crossentropy",
              metrics=["accuracy"])
model.summary()   # печатает структуру сети и число весов
```

_Ожидаемый результат: таблица со слоями сети и числом обучаемых параметров (весов)._

Метод .compile() задаёт, как сеть будет учиться: optimizer — способ подкручивать веса (adam — надёжный выбор по умолчанию), loss — как измерять ошибку. Обучение запускается методом .fit(), как и в sklearn.

**Внимание, частая ошибка  ⚠️**

activation="relu" — функция активации, которая позволяет сети улавливать сложные, нелинейные закономерности. Без неё сеть из многих слоёв схлопывается в один простой слой и теряет всю мощь. Частая ошибка новичков — забыть активацию в скрытых слоях. Пока запомни простое правило: relu в скрытых слоях почти всегда, а в выходном — sigmoid (для да/нет) или softmax (для нескольких классов).
