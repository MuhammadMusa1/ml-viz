## Шаг 1. Интерфейс средствами Gradio

С Gradio мы уже встречались в главе 16. Напомним суть: ты пишешь обычную Python-функцию, а Gradio строит вокруг неё веб-страницу с полями ввода и вывода. Тебе не нужно знать HTML, CSS или веб-программирование — Gradio берёт это на себя. Соберём простое приложение-калькулятор на модели регрессии «часы → баллы» из главы 8.

```python
# Запускать в Google Colab
!pip install gradio -q

import gradio as gr
import numpy as np
from sklearn.linear_model import LinearRegression

# Обучаем ту же модель "часы учёбы -> баллы" из главы 8
hours = np.array([1,2,3,4,5,6,7,8]).reshape(-1,1)
scores = np.array([20,35,40,55,60,72,80,88])
model = LinearRegression().fit(hours, scores)

def predict(h):
    return round(float(model.predict([[h]])[0]), 1)

# Интерфейс: ползунок на вход, число на выход
demo = gr.Interface(
    fn=predict,
    inputs=gr.Slider(0, 12, label="Часов учёбы"),
    outputs=gr.Number(label="Прогноз баллов"),
    title="Предсказатель оценок",
)
demo.launch(share=True)   # share=True — временная публичная ссылка
```

_Ожидаемый результат: рабочее приложение с ползунком + ссылка *.gradio.live на 72 часа._

Обрати внимание, как просто устроен Gradio: у функции predict есть вход (число часов) и выход (баллы), а Gradio сам делает под них ползунок и поле результата. Меняешь тип ввода на gr.Image() — и получаешь загрузчик картинок, как в проекте главы 16.

**Полезно знать  ℹ️**

Ссылка от share=True временная — она живёт около 72 часов и работает, только пока запущен твой блокнот Colab. Закрыл блокнот — ссылка умерла. Для постоянного приложения, которое доступно всегда и не зависит от твоего компьютера, нужен настоящий хостинг. Им и займёмся дальше.

## Шаг 2. Постоянный дом — Hugging Face Spaces

Hugging Face Spaces — бесплатный хостинг для ML-приложений. Ты загружаешь туда код, а Spaces сам запускает его на своём сервере и даёт постоянный адрес. Приложение работает круглосуточно, даже когда твой компьютер выключен, а ноутбук лежит в рюкзаке. Именно так выкладывают демо настоящие ML-инженеры и исследователи.

### Как выложить приложение — по шагам

Зарегистрируйся на huggingface.co (бесплатно, нужна только почта).

Нажми «New Space», выбери SDK «Gradio» и придумай название своему приложению.

Положи код приложения в файл app.py.

Перечисли нужные библиотеки в файле requirements.txt (например, gradio, scikit-learn).

Загрузи оба файла в Space — он сам всё соберёт, установит библиотеки и запустит.

## Что положить в app.py

Файл app.py — это тот же код Gradio, но без share=True: на Spaces публичный адрес даётся автоматически, временная ссылка не нужна. Вот минимальный рабочий пример.

```python
# app.py — файл для Hugging Face Spaces
import gradio as gr
import numpy as np
from sklearn.linear_model import LinearRegression

hours = np.array([1,2,3,4,5,6,7,8]).reshape(-1,1)
scores = np.array([20,35,40,55,60,72,80,88])
model = LinearRegression().fit(hours, scores)

def predict(h):
    return round(float(model.predict([[h]])[0]), 1)

demo = gr.Interface(fn=predict,
                    inputs=gr.Slider(0, 12, label="Часов учёбы"),
                    outputs=gr.Number(label="Прогноз баллов"))
demo.launch()   # на Spaces адрес выдаётся сам, share не нужен
# requirements.txt — список библиотек, которые Spaces установит сам
gradio
scikit-learn
numpy
```

_Два файла — и приложение живёт по постоянному адресу вида huggingface.co/spaces/твоё-имя/название._

**Внимание  ⚠️**

Регистрация, загрузка файлов и работа Spaces происходят на сайте huggingface.co и требуют интернета — это делается не в среде без сети. Код app.py удобно сначала написать и проверить в Colab (там всё работает), а затем перенести на Spaces. Если приложение на Spaces не запускается — почти всегда дело в забытой библиотеке в requirements.txt.
