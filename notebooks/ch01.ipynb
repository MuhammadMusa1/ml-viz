## Две разные идеи: правила и обучение

Представь, что тебя попросили написать программу, которая отличает кошку от собаки на фотографии. Как бы ты это сделал обычным способом — через правила?

Ты бы начал: «Если уши треугольные и торчат вверх — это кошка. Если морда вытянутая — собака…» И очень быстро увяз бы. А вислоухая кошка? А чихуахуа с торчащими ушами? А фото сбоку, в темноте, с одной видной лапой? Правил становятся тысячи, они противоречат друг другу и всё равно не работают. Ты потратишь месяцы и не добьёшься результата.

_Машинное обучение переворачивает задачу с ног на голову. Вместо того чтобы писать правила, ты показываешь машине тысячи примеров — «вот кошка», «вот собака», «вот снова кошка» — и она сама находит закономерности, которые отличают одно от другого. Правила она выводит без тебя, прямо из примеров._

**Аналогия: рецепт и дегустатор  💡**

Обычное программирование — это рецепт: ты записываешь точные шаги, и повар их выполняет буква в букву. Машинное обучение — это дегустатор, который попробовал сотни блюд и теперь на вкус угадывает, солёное блюдо или сладкое, хотя рецепта ему никто не давал. Он научился на опыте, а не по инструкции.

Разница огромна. В мире правил программист должен заранее предусмотреть каждый случай. В мире обучения программист лишь готовит хорошие примеры, а закономерности машина находит сама — включая те, которые человек и не заметил бы.

### Определение своими словами

**Главная мысль  🎯**

Машинное обучение — это способ научить компьютер решать задачу не через записанные вручную правила, а через примеры, из которых он сам выводит закономерности (паттерны).

## Что такое «паттерн в данных»

Паттерн — это устойчивая закономерность, повторяющаяся связь между явлениями. Люди находят паттерны постоянно, даже не замечая этого. Ты знаешь, что после серых туч часто идёт дождь. Никто не давал тебе формулу «туча → дождь» — ты вывел её сам из множества наблюдений за небом. Именно этим занимается модель машинного обучения, только с числами и в огромных количествах.

Например, в данных о жилье может прятаться паттерн: «чем больше площадь, тем выше цена». В данных о фильмах: «фильмы с рейтингом выше 8 чаще собирают большую кассу». В данных о письмах: «письма со словами „выигрыш“ и „срочно“ чаще оказываются спамом». Модель ищет такие связи и потом использует их, чтобы делать предсказания на новых, ещё не виденных данных.

Ключевое слово — «предсказание». Модель, нашедшая паттерн на старых данных, применяет его к новым. Увидев новое письмо, она предскажет: спам или нет. Увидев новую квартиру, предскажет цену. В этом вся суть.

## Типы задач машинного обучения

Почти всё, что мы будем делать в этой книге, сводится к трём большим типам задач. Разберём каждый на понятном примере, а потом сведём в таблицу.

### 1. Классификация — «к какой категории?»

Модель выбирает одну из нескольких заранее известных категорий. Спам или не спам. Кошка, собака или птица. Положительный отзыв или отрицательный. Сдаст ученик экзамен или нет. Ответ — это метка (label), одна из фиксированного набора вариантов. Категорий может быть две (тогда это бинарная классификация) или много.

### 2. Регрессия — «сколько?»

Модель предсказывает число на непрерывной шкале. Какой будет цена квартиры. Сколько баллов наберёт ученик. Какая температура будет завтра. Сколько зрителей соберёт фильм. Ответ — это число, и оно может принимать любое значение из диапазона, а не только несколько фиксированных вариантов.

### 3. Кластеризация — «кто на кого похож?»

Модель сама, без подсказок, разбивает данные на группы похожих объектов. Например, находит среди покупателей магазина группы со схожим поведением, чтобы предложить каждой своё. Здесь принципиальное отличие: никто заранее не говорит модели правильный ответ — она группирует данные сама, опираясь только на их внутреннюю структуру.

**Внимание, частая ошибка  ⚠️**

Классификацию и регрессию путают чаще всего. Запомни разницу по типу ответа: если ответ — это категория (одна из нескольких), это классификация. Если ответ — число на непрерывной шкале, это регрессия. «Сдаст экзамен / не сдаст» — классификация (две категории). «Сколько баллов наберёт» — регрессия (число от 0 до 100). Один и тот же экзамен — но две совершенно разные задачи!

## Обучение с учителем и без учителя

Три типа задач делятся на две большие группы по важному признаку — есть ли в данных правильные ответы для обучения.

Обучение с учителем (supervised) — в данных есть правильные ответы, на которых модель тренируется. Классификация и регрессия относятся сюда: мы показываем модели письма с пометками «спам/не спам» или квартиры с известными ценами. «Учитель» — это как раз готовые правильные ответы, по которым модель сверяется.

Обучение без учителя (unsupervised) — правильных ответов нет, модель ищет структуру сама. Кластеризация относится сюда. В этом курсе мы почти всё время работаем с обучением с учителем — оно нагляднее и понятнее в начале пути, а результат легко проверить.

## Где ML уже работает вокруг тебя

Машинное обучение — не абстракция из будущего. Оно уже встроено в твою повседневную жизнь, чаще всего незаметно:

Лента рекомендаций в YouTube и TikTok — модель предсказывает, какое видео тебя удержит.

Автодополнение и исправление в клавиатуре телефона — предсказывает следующее слово.

Распознавание лица при разблокировке телефона — классификация «это владелец или нет».

Фильтр спама в почте — классификация писем на нужные и мусорные.

Голосовые помощники — распознают речь и понимают команды.

Каждый из этих примеров — задача, которую невозможно решить набором ручных правил, но легко через обучение на данных. К концу книги ты будешь понимать, как устроен каждый из них, и сможешь сделать что-то похожее сам.

## Попробуй сам: игра «Угадай правило»

Эту игру можно сыграть без единой строчки кода — с другом или всем классом. Она показывает суть машинного обучения буквально за пять минут и запоминается лучше любого определения.

Один игрок загадывает секретное правило, например: «число проходит, если оно чётное». Остальным правило не говорят.

Второй игрок называет числа наугад: 4 → «проходит», 7 → «не проходит», 10 → «проходит», 3 → «не проходит»…

С каждым ответом второй игрок уточняет свою гипотезу о правиле — ровно как модель уточняет закономерность на новых примерах.

Когда второй уверен — он называет правило вслух. Угадал? Значит «обучение» прошло успешно.

Ты только что сработал как модель машинного обучения: получал примеры с ответами («проходит»/«не проходит») и постепенно выводил из них скрытое правило. Именно это делает алгоритм — только с тысячами примеров и очень быстро. А ещё обрати внимание: чем больше примеров ты получил, тем увереннее твоя догадка. У моделей точно так же — больше данных обычно означает точнее предсказание.
