До сих пор мы предсказывали категории — выжил или нет. Но огромный класс задач требует предсказать число: цену квартиры, завтрашнюю температуру, оценку за экзамен, кассовые сборы фильма. Это регрессия — второй кит машинного обучения наряду с классификацией.

**Классификация против регрессии  ℹ️**

Классификация отвечает на вопрос «какая категория?» (кошка/собака, спам/не спам). Регрессия отвечает на вопрос «сколько?» (сколько будет стоить, сколько баллов). Если ответ — число на непрерывной шкале, это регрессия. Один признак может служить обеим задачам: «сдаст ли экзамен» — классификация, «сколько баллов наберёт» — регрессия.

## Линейная регрессия: линия через облако точек

**Аналогия: линейка через точки  💡**

Нанеси на график точки: по горизонтали — часы учёбы, по вертикали — оценка за экзамен. Точки образуют «облако», уходящее вверх: больше учишься — выше оценка. Линейная регрессия проводит через это облако прямую линию так, чтобы она проходила максимально близко ко всем точкам сразу. По этой линии можно предсказать оценку для любого числа часов — даже такого, которого не было в данных.

Как модель находит «лучшую» линию? Она подбирает такой наклон и высоту прямой, чтобы суммарное расстояние от точек до линии было минимальным. Линия, максимально близкая ко всем точкам, и есть найденная закономерность.

## Регрессия в коде на синтетических данных

Возьмём простые придуманные данные «часы учёбы → баллы» — они идеально показывают идею, потому что связь в них очевидна и наглядна.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Синтетические данные: часы учёбы и баллы
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
scores = np.array([20, 35, 40, 55, 60, 72, 80, 88])

model = LinearRegression()
model.fit(hours, scores)          # находим лучшую прямую

# Рисуем точки и линию модели
plt.scatter(hours, scores, label="реальные данные")
plt.plot(hours, model.predict(hours), color="red", label="линия модели")
plt.xlabel("Часы учёбы"); plt.ylabel("Баллы"); plt.legend()
plt.show()

# Предсказание: сколько баллов при 5.5 часах учёбы?
print("Прогноз:", model.predict([[5.5]]))
```

_Ожидаемый результат: график с точками и красной линией + прогноз около 66 баллов._

Метод .reshape(-1, 1) превращает список чисел в столбец — sklearn ожидает признаки именно в форме таблицы (каждая строка — объект). Это частый источник ошибок у новичков, так что запомни: признаки всегда двумерные.

## Метрики регрессии: MAE и MSE

В классификации мы считали долю правильных ответов (accuracy). В регрессии ответ — число, и «правильно/неправильно» не подходит: важно, насколько сильно мы промахнулись. Для этого есть две главные метрики.

Метрика

Что это

Особенность

MAE

Средняя абсолютная ошибка — на сколько в среднем промахиваемся

В единицах задачи (баллы, рубли). Легко понять и объяснить

MSE

Средний квадрат ошибки

Сильно наказывает крупные промахи из-за возведения в квадрат

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

pred = model.predict(hours)
print("MAE:", mean_absolute_error(scores, pred))
print("MSE:", mean_squared_error(scores, pred))
```

**Полезно знать  ℹ️**

MAE = 4 означает: в среднем модель ошибается на 4 балла. Это интуитивно понятно, потому что измеряется в тех же единицах, что и цель. MSE труднее толковать напрямую (баллы в квадрате?), зато она сильнее реагирует на большие ошибки — полезно, когда крупные промахи особенно нежелательны (например, ошибиться в дозировке лекарства на 10 хуже, чем дважды на 5).
