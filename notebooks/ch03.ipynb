## Три графика, которые нужны почти всегда

Существуют десятки типов графиков, но для старта достаточно трёх. Разберём каждый: что он показывает, когда применять и как читать.

### Гистограмма — как распределены значения одного признака

Гистограмма показывает, каких значений в данных много, а каких мало. Она делит диапазон на интервалы («столбики») и считает, сколько объектов попало в каждый. Например, какой длины ласты среди пингвинов встречаются чаще всего.

```python
# Данные о пингвинах (выполни, если открыл новый блокнот)
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url).drop(columns=["sex"])

import matplotlib.pyplot as plt   # главная библиотека для графиков

# Гистограмма длины ласта пингвинов
df["flipper_length_mm"].hist(bins=20)   # bins — на сколько столбиков делить
plt.xlabel("Длина ласта, мм")     # подпись оси X
plt.ylabel("Сколько пингвинов")   # подпись оси Y
plt.title("Распределение длины ласта")
plt.show()
```

_Ожидаемый результат: столбчатая диаграмма — видно, что чаще всего встречаются ласты длиной 190–200 мм._

Как читать: высокий столбик означает «таких значений много». Если гистограмма имеет один «горб» — данные сосредоточены вокруг типичного значения. Два горба могут намекать на две разные группы внутри данных. Параметр bins стоит поменять (10, 20, 50) и посмотреть, как меняется картина.

### Диаграмма рассеяния (scatter plot) — связь двух признаков

Каждая точка — один объект. По горизонтальной оси один признак, по вертикальной — второй. Так видно, связаны ли они между собой. Если точки складываются в линию или облако с наклоном — связь есть.

```python
# Есть ли связь между длиной ласта и массой тела?
df.plot.scatter(x="flipper_length_mm", y="body_mass_g")
plt.title("Длина ласта против массы тела")
plt.show()
```

Как читать: если облако точек идёт снизу вверх слева направо — чем больше один признак, тем больше другой (положительная связь). Если сверху вниз — наоборот (отрицательная связь). Если точки разбросаны бесформенным пятном — связи нет. Диаграмма рассеяния особенно полезна перед регрессией, о которой в фазе 2.

### Тепловая карта корреляций (heatmap) — что с чем связано

Показывает сразу для всех пар числовых признаков, насколько сильно они связаны. Понадобится Seaborn — библиотека красивых графиков, построенная поверх Matplotlib.

```python
import seaborn as sns

# Считаем корреляции только по числовым колонкам
corr = df.corr(numeric_only=True)

# Рисуем тепловую карту: чем ярче/теплее клетка, тем сильнее связь
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Корреляции между признаками")
plt.show()
```

Параметр annot=True подписывает в каждой клетке точное число корреляции. Это удобно: видно и цветом, и цифрой. Ищи самые тёмные и самые яркие клетки — это самые сильные связи в данных.

## Что такое корреляция

**Аналогия: качели  💡**

Корреляция — это число от −1 до +1, показывающее, как связаны два признака. +1: когда один растёт, другой тоже растёт (как рост и вес человека). −1: когда один растёт, другой падает (как высота горы и температура на её вершине: чем выше, тем холоднее). Около 0: связи нет, признаки ведут себя независимо, как качели, стоящие в разных дворах.

Сила корреляции — это её удалённость от нуля. Корреляция +0,9 — очень сильная прямая связь. −0,8 — сильная обратная. +0,1 — почти никакой связи. Знак (плюс или минус) говорит о направлении, а величина — о силе.

**Внимание, частая ошибка  ⚠️**

Корреляция — это НЕ причина! Если продажи мороженого и число солнечных ожогов растут вместе (высокая корреляция), это не значит, что мороженое вызывает ожоги. Просто оба явления зависят от третьего — жаркой погоды. Всегда спрашивай себя: нет ли скрытой третьей причины? Путать корреляцию с причинно-следственной связью — одна из самых частых и опасных ошибок в анализе данных.

## Главное правило EDA: график без вывода бесполезен

Построить график — только половина дела. Вторая половина, и самая важная, — сформулировать словами, что он показывает. График, под которым ничего не написано, никому не помогает. Возьми за железное правило: под каждым графиком писать вывод одним-двумя предложениями. Например: «На гистограмме видно, что у большинства пингвинов ласты длиной 190–200 мм, а очень коротких и очень длинных мало».

Этот навык — превращать картинку в словесный вывод — отличает того, кто просто рисует графики, от того, кто понимает данные. На защите проектов в фазе 6 именно за него ставят баллы.
