# Глава 24. Demo Day: презентация и защита

_Ноутбук сгенерирован из учебника «Машинное обучение для начинающих». Запускайте ячейки по порядку сверху вниз в Google Colab._

**Цель главы  🎯**

После этой главы ты сможешь уверенно защитить свой проект перед аудиторией, ответить на вопросы и оценить свою работу по критериям защиты.

Финальный день курса. Приходят родители и, возможно, гости из IT. Каждый ученик показывает своё работающее приложение и защищает проект. Это не экзамен, которого нужно бояться, а праздник — момент, когда ты показываешь то, что создал своими руками за три месяца. Волноваться нормально; помни, что зал настроен доброжелательно и хочет увидеть твой успех.

## Как проходит защита

Ты выступаешь по отрепетированной структуре из главы 23, показываешь живое приложение и отвечаешь на вопросы. Вопросы — не попытка завалить, а искренний интерес к твоей работе. Отвечай спокойно и честно. Фраза «я это не реализовал, но знаю, как бы сделал» — тоже хороший, зрелый ответ: он показывает понимание, даже если не всё успел воплотить.

**Полезно знать  ℹ️**

Если приложение вдруг не запустилось или интернет подвёл — не паникуй. Спокойно покажи запасные скриншоты или запись работы (которые ты подготовил после главы 23) и продолжай. Умение спокойно справиться с накладкой производит даже лучшее впечатление, чем гладкое выступление. Технические сбои случаются у всех, включая профессионалов.

## Как оценивается защита

Оценка — это не отметка ради отметки, а признание проделанной работы и обратная связь для роста. Рубрика проста и известна заранее, чтобы ты понимал, на что обратить внимание.

**Результат защиты  ℹ️**

Максимум — 12 баллов. Сертификат с отличием — от 10 баллов, сертификат о прохождении курса — от 6 баллов. Но помни: главная награда не в баллах, а в ссылке на живой продукт, который ты сделал сам и можешь показать кому угодно. Баллы забудутся, а работающее приложение и уверенность в своих силах останутся.

## Что ты умеешь теперь

**Ты прошёл весь путь  🎯**

Ты начал с вопроса «чем ML отличается от программирования» — и дошёл до собственного развёрнутого приложения в интернете. Ты умеешь читать и чистить данные, строить и честно оценивать модели, обучать нейросети, работать с LLM через промпты, ускоряться через AutoML и выкладывать продукты в интернет. И, что важнее всего, ты можешь объяснить, что делаешь, — своими словами, понятно для любого. Это фундамент настоящего специалиста по ИИ.

## Задания

Базовый уровень

Защити свой проект на Demo Day по отрепетированной структуре. После защиты честно оцени себя по рубрике из четырёх критериев: где ты силён, а что стоит подтянуть в следующем проекте?

Со звёздочкой  ⭐

После Demo Day напиши короткую саморефлексию: что получилось лучше всего, какая часть далась труднее всего и какой проект ты хотел бы сделать следующим, применив полученные навыки. Это твой личный план роста.

Вопросы для обсуждения  💬

Что производит большее впечатление на защите — высокая точность модели или ясный рассказ и живое демо? Почему?

Как лучше ответить на вопрос, ответа на который ты не знаешь?

Чему из курса ты научился, что пригодится даже вне машинного обучения?

**Проверь себя  ℹ️**

1. Как вести себя, если приложение не запустилось во время защиты?

2. По каким четырём критериям оценивается проект?

3. Что для тебя главная награда за проект?
