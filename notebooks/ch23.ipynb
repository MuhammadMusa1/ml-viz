# Глава 23. Финальный проект: репетиция

_Ноутбук сгенерирован из учебника «Машинное обучение для начинающих». Запускайте ячейки по порядку сверху вниз в Google Colab._

**Цель главы  🎯**

После этой главы ты сможешь понятно рассказать о проекте нетехническому слушателю, выстроить защиту по чёткой структуре на 5 минут и проверить готовность по финальному чек-листу.

Проект почти готов. Осталось научиться его рассказывать — и это отдельный важный навык, который многие недооценивают. Лучшая модель бесполезна, если автор не может объяснить, что она делает и почему это ценно. Эта глава — про репетицию защиты, потому что хорошая защита не рождается сама, её репетируют.

## Рассказывай понятным языком

**Аналогия: объясни бабушке  💡**

Представь, что рассказываешь проект человеку, который никогда не слышал слов «Pipeline», «accuracy» и «нейросеть» — например, бабушке или младшему брату. Если получилось объяснить ему, что делает твоё приложение и почему это круто, — значит, ты понял проект по-настоящему. Технические термины прячь за простыми словами: не «классификатор на RandomForest с F1 = 0,82», а «программа, которая по описанию угадывает жанр фильма и почти всегда попадает».

## Структура защиты на 5 минут

Хорошая защита следует ясной структуре — она помогает и тебе не сбиться, и слушателям следить за мыслью. Отрепетируй вслух и уложись в 5 минут; лучше меньше, но чётко, чем долго и сумбурно.

Проблема. Какую задачу решаешь и почему она интересна. (30 секунд)

Данные. С чем работал, откуда взял, что в них было. (1 минута)

Решение. Как построил модель, какую метрику выбрал и почему, как сравнил с baseline. (1,5 минуты)

Демонстрация. Живой показ приложения — самая важная и эффектная часть! (1,5 минуты)

Вывод. Что получилось, чему научился, что можно улучшить дальше. (30 секунд)

**Полезно знать  ℹ️**

Живая демонстрация — сердце защиты. Ничто не производит такого впечатления, как работающее приложение, в которое прямо на глазах у зрителей вводят данные и получают ответ. Обязательно проверь заранее, что приложение открывается и работает, — а лучше подготовь запасной вариант (скриншоты или запись), если вдруг подведёт интернет.

## Обратная связь от группы

Прогони защиту перед одноклассниками до Demo Day. Пусть каждый задаст один вопрос и назовёт одну вещь, которую стоит улучшить. Чужой взгляд ловит то, что автору не видно: непонятный термин, неудобную кнопку в приложении, слабое место в рассуждении. Это репетиция настоящих вопросов, которые прозвучат на защите, — и шанс подготовить ответы заранее.

## Финальный чек-лист проекта

Проект считается завершённым, только если выполнены все обязательные пункты. Пройди по списку честно — это ровно те критерии, по которым проект оценивается на Demo Day.

**Чек-лист готовности  🎯**

Задача сформулирована и реализована целиком (end-to-end).

Используется Pipeline для предобработки и модели.

Есть метрика качества и сравнение с baseline.

Есть работающий веб-интерфейс и публичная ссылка.

Есть README с описанием задачи, данных и запуска.

Ты можешь защитить проект и ответить на вопросы.

## Задания

Базовый уровень

Отрепетируй защиту вслух по пятишаговой структуре, засекая время. Уложился в 5 минут? Затем расскажи проект кому-то нетехническому (родителю, другу) и проверь: понял ли он, что делает приложение?

Со звёздочкой  ⭐

Проведи полную репетицию перед группой: выступи, собери по одному вопросу и одному замечанию от каждого. Подготовь ответы на все вопросы и внеси хотя бы одно улучшение по замечаниям. Пройди финальный чек-лист — все ли пункты выполнены?

Вопросы для обсуждения  💬

Зачем объяснять проект «языком для бабушки», если защищаешь его перед знающими преподавателями?

Почему живая демонстрация важнее красивых слайдов с цифрами?

Какой пункт чек-листа обычно забывают в последний момент? Как этого избежать?

**Проверь себя  ℹ️**

1. Зачем объяснять проект простым языком?

2. Какая часть защиты самая важная и почему?

3. Все ли пункты чек-листа у тебя выполнены?
