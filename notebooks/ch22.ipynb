## Улучшаем модель — но честно

Улучшение модели — это не «сделать число побольше любой ценой». Это осмысленные шаги, каждый из которых проверяется по кросс-валидации (глава 10), чтобы не обмануть себя случайным везением на одном разбиении.

Добавь или преобразуй признаки (feature engineering) — часто это даёт больше, чем смена модели. Новый осмысленный признак ценнее нового алгоритма.

Попробуй разные модели: лес, градиентный бустинг, регрессию — и сравни честно по кросс-валидации.

Подбери параметры через GridSearchCV, если данных немного и времени хватает.

Сравнивай каждый шаг с baseline и с предыдущей версией: стало реально лучше или это шум?

**Внимание: осторожно с утечкой  ⚠️**

Улучшая модель, легко случайно устроить утечку данных (глава 6) — например, посчитать масштабирование или заполнение пропусков на всех данных до разделения. Держи ВСЕ преобразования внутри Pipeline: это и есть твоя автоматическая защита от утечки. Если вдруг точность подскочила до подозрительно высокой (0,99 там, где ждал 0,80) — первым делом ищи утечку, а не радуйся.

## Выбери правильную метрику

Вспомни главу 11: accuracy подходит не всегда, особенно на несбалансированных данных. Задай себе вопрос, что дороже — ложная тревога или пропуск, — и выбери метрику под задачу (precision, recall, F1 или, для регрессии, MAE/MSE). На защите нужно будет объяснить, почему выбрана именно эта метрика, а не взята accuracy «по умолчанию». Осознанный выбор метрики — признак зрелого специалиста.
